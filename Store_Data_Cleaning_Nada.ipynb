# change the the date format

```python
df['Store_Open_Date'] = pd.to_datetime(df['Store_Open_Date'])
print(df['Store_Open_Date'].head())
```

```python
print(df.dtypes)
```

```python
df['Store_Name'] = df['Store_Name'].str.strip()
df['Store_City'] = df['Store_City'].str.strip()
df['Store_Location'] = df['Store_Location'].str.strip()


# Convert to lowercase
df['Store_Name'] = df['Store_Name'].str.lower()
df['Store_City'] = df['Store_City'].str.lower()
df['Store_Location'] = df['Store_Location'].str.lower()


# Convert to title case (First Letter Capitalized)
df['Store_Name'] = df['Store_Name'].str.title()
df['Store_City'] = df['Store_City'].str.title()
df['Store_Location'] = df['Store_Location'].str.title()
```

```python
#check the null values
print(df.isna().sum())
```
